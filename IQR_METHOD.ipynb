```python
import pandas as pd

# Dataset load karo
df = pd.read_csv("IQR_Outlier_Practice_Dataset.csv")

print(df.shape)
# Salary column ke liye Q1 and Q3
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

# IQR calculate
IQR = Q3 - Q1

# Limits calculate
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Outliers remove
df = df[
    (df["Salary"] >= lower_limit) &
    (df["Salary"] <= upper_limit)
]

# Final data
print(df)
```
