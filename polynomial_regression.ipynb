```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = pd.read_csv("Position_Level_Salary.csv")
```

```python
X = df[["Level"]]
y = df["Salary"]
```

```python
lr = LinearRegression()
lr.fit(X, y)
```

```python
poly = PolynomialFeatures(degree=4)

X_poly = poly.fit_transform(X)
```

```python
pr = LinearRegression()
pr.fit(X_poly, y)
```

```python
plt.scatter(X, y, color="red")
plt.plot(X, lr.predict(X), color="blue")
plt.title("Linear Regression")
plt.xlabel("Level")
plt.ylabel("Salary")
plt.show()
```

```python
plt.scatter(X, y, color="red")
plt.plot(X, pr.predict(X_poly), color="green")
plt.title("Polynomial Regression")
plt.xlabel("Level")
plt.ylabel("Salary")
plt.show()
```

```python
level = [[6.5]]
```

```python
linear_salary = lr.predict(level)
```

```python
poly_salary = pr.predict(poly.transform(level))
```

```python
print("Linear Prediction:", linear_salary)
print("Polynomial Prediction:", poly_salary)
```
