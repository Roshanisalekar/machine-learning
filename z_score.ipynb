```python
import pandas as pd
from scipy.stats import zscore

# Dataset load karo
df = pd.read_csv("IQR_Outlier_Practice_Dataset.csv")

# Salary ka Z-score calculate karo
df["Z_Score"] = zscore(df["Salary"])

# Outliers remove karo
df = df[
    (df["Z_Score"] >= -3) &
    (df["Z_Score"] <= 3)
]

# Z_Score column remove karo
df = df.drop("Z_Score", axis=1)

# Final data print karo
print(df)
```
