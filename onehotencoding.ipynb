```python
# 1. Import libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import OneHotEncoder


# 2. Load dataset
df = pd.read_csv("Employee_Performance_EDA_Practice.csv")

# Check original data
print(df.head())


# 3. Select categorical columns
en_data = df[["Gender", "Promotion", "Purchased"]]


# 4. Create One-Hot Encoder
ohe = OneHotEncoder(drop="first", sparse_output=False)


# 5. Fit and transform data
ar = ohe.fit_transform(en_data)


# 6. Create DataFrame with encoded columns
encoded_df = pd.DataFrame(
    ar,
    columns=ohe.get_feature_names_out(en_data.columns),
    index=df.index
)


# 7. Remove original text columns
df = df.drop(
    ["Gender", "Promotion", "Purchased"],
    axis=1
)


# 8. Add encoded columns to original DataFrame
df = pd.concat(
    [df, encoded_df],
    axis=1
)


# 9. Check final data
print(df.head())
print(df.info())
```
